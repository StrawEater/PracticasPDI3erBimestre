```python
# @title
# ============================================================
# 🔬 LABORATORIO INTERACTIVO DE FOURIER
# ============================================================
# Dibujá en el ESPACIO ORIGINAL y observá su TRANSFORMADA
# DE FOURIER en tiempo real.
#
# Copiá todo en UNA SOLA CELDA de Google Colab.
# No requiere instalar nada extra.
# ============================================================

from IPython.display import HTML, display

display(HTML("""
<style>
  #fourier-lab *, #fourier-lab *::before, #fourier-lab *::after {
    box-sizing: border-box;
    margin: 0;
    padding: 0;
  }

  #fourier-lab {
    font-family: 'Segoe UI', system-ui, -apple-system, sans-serif;
    background: linear-gradient(145deg, #0d1117 0%, #161b22 50%, #0d1117 100%);
    color: #e6edf3;
    padding: 28px;
    border-radius: 18px;
    box-shadow: 0 8px 40px rgba(0,0,0,0.5);
    max-width: 920px;
    margin: 10px auto;
    border: 1px solid #21262d;
  }

  .fl-header {
    text-align: center;
    margin-bottom: 22px;
  }

  .fl-header h2 {
    font-size: 22px;
    font-weight: 700;
    background: linear-gradient(90deg, #58a6ff, #79c0ff);
    -webkit-background-clip: text;
    -webkit-text-fill-color: transparent;
    background-clip: text;
    margin-bottom: 6px;
  }

  .fl-header p {
    font-size: 13px;
    color: #8b949e;
  }

  .fl-canvases {
    display: flex;
    justify-content: center;
    align-items: flex-start;
    gap: 16px;
    flex-wrap: wrap;
    margin-bottom: 20px;
  }

  .fl-panel {
    background: rgba(22,27,34,0.85);
    border: 1px solid #30363d;
    border-radius: 14px;
    padding: 14px;
    text-align: center;
    transition: border-color 0.3s;
  }

  .fl-panel:hover {
    border-color: #484f58;
  }

  .fl-panel-title {
    font-size: 13px;
    font-weight: 600;
    margin-bottom: 10px;
    display: flex;
    align-items: center;
    justify-content: center;
    gap: 8px;
  }

  .fl-dot {
    width: 8px;
    height: 8px;
    border-radius: 50%;
    display: inline-block;
  }

  .fl-dot.blue {
    background: #58a6ff;
    box-shadow: 0 0 6px #58a6ff;
  }

  .fl-dot.amber {
    background: #d29922;
    box-shadow: 0 0 6px #d29922;
  }

  .fl-panel canvas {
    width: 340px;
    height: 340px;
    display: block;
    margin: 0 auto;
    border-radius: 8px;
    border: 2px solid #30363d;
    image-rendering: pixelated;
    image-rendering: -moz-crisp-edges;
    image-rendering: crisp-edges;
    background: #000;
  }

  .fourier-panel canvas {
    box-shadow: 0 0 18px rgba(56,189,248,0.1);
  }

  .spatial-panel canvas {
    cursor: crosshair;
    box-shadow: 0 0 18px rgba(210,153,34,0.1);
  }

  .fl-panel-sub {
    font-size: 11px;
    color: #8b949e;
    margin-top: 8px;
  }

  .fl-arrow {
    display: flex;
    flex-direction: column;
    align-items: center;
    justify-content: center;
    padding-top: 60px;
    gap: 4px;
  }

  .fl-arrow-label {
    font-size: 11px;
    color: #58a6ff;
    font-weight: 600;
    letter-spacing: 1px;
    text-transform: uppercase;
  }

  .fl-arrow-icon {
    font-size: 28px;
    color: #58a6ff;
    animation: fl-pulse 2s ease-in-out infinite;
  }

  @keyframes fl-pulse {
    0%, 100% {
      opacity: 0.5;
      transform: translateX(0);
    }

    50% {
      opacity: 1;
      transform: translateX(4px);
    }
  }

  .fl-toolbar {
    display: flex;
    justify-content: center;
    align-items: center;
    gap: 16px;
    flex-wrap: wrap;
    margin-bottom: 14px;
    padding: 12px 16px;
    background: rgba(22,27,34,0.7);
    border-radius: 10px;
    border: 1px solid #21262d;
  }

  .fl-tool-group {
    display: flex;
    align-items: center;
    gap: 8px;
  }

  .fl-btn {
    background: #21262d;
    color: #e6edf3;
    border: 1px solid #30363d;
    border-radius: 8px;
    padding: 7px 14px;
    cursor: pointer;
    font-size: 12px;
    transition: all 0.2s ease;
    font-family: inherit;
    user-select: none;
    white-space: nowrap;
  }

  .fl-btn:hover {
    background: #30363d;
    border-color: #58a6ff;
  }

  .fl-btn.active {
    background: #1f6feb;
    border-color: #58a6ff;
    color: #fff;
  }

  .fl-label {
    font-size: 12px;
    color: #8b949e;
    display: flex;
    align-items: center;
    gap: 6px;
  }

  .fl-label input[type=range] {
    width: 90px;
    accent-color: #58a6ff;
    cursor: pointer;
  }

  .fl-label span {
    color: #e6edf3;
    font-weight: 600;
    min-width: 18px;
    text-align: center;
  }

  .fl-presets {
    display: flex;
    justify-content: center;
    align-items: center;
    gap: 8px;
    flex-wrap: wrap;
  }

  .fl-presets-label {
    font-size: 12px;
    color: #8b949e;
  }

  .fl-preset {
    padding: 5px 12px;
    font-size: 11px;
    background: #161b22;
    border-color: #21262d;
  }

  .fl-preset:hover {
    background: #21262d;
    border-color: #484f58;
  }

  @media (max-width: 760px) {
    .fl-panel canvas {
      width: 280px;
      height: 280px;
    }

    .fl-arrow {
      padding-top: 10px;
    }

    .fl-arrow-icon {
      transform: rotate(90deg);
    }

    #fourier-lab {
      padding: 16px;
    }
  }
</style>


<div id="fourier-lab">

  <div class="fl-header">
    <h2>🔬 Laboratorio Interactivo de Fourier</h2>
    <p>
      Dibujá en el espacio original y observá su Transformada de Fourier
      en tiempo real
    </p>
  </div>


  <div class="fl-canvases">

    <!-- =====================================================
         ESPACIO ORIGINAL
         ===================================================== -->

    <div class="fl-panel spatial-panel">

      <div class="fl-panel-title">
        <span class="fl-dot amber"></span>
        Espacio Original
      </div>

      <canvas
        id="fl-spatial"
        width="256"
        height="256">
      </canvas>

      <div class="fl-panel-sub">
        ✏️ Dibujá aquí con el mouse
      </div>

    </div>


    <!-- =====================================================
         FLECHA
         ===================================================== -->

    <div class="fl-arrow">

      <div class="fl-arrow-label">
        FFT
      </div>

      <div class="fl-arrow-icon">
        →
      </div>

    </div>


    <!-- =====================================================
         ESPACIO DE FOURIER
         ===================================================== -->

    <div class="fl-panel fourier-panel">

      <div class="fl-panel-title">
        <span class="fl-dot blue"></span>
        Espacio de Fourier
      </div>

      <canvas
        id="fl-fourier"
        width="256"
        height="256">
      </canvas>

      <div class="fl-panel-sub">
        Magnitud de la FFT (centrada)
      </div>

    </div>

  </div>


  <!-- =======================================================
       CONTROLES
       ======================================================= -->

  <div class="fl-toolbar">

    <div class="fl-tool-group">

      <button class="fl-btn active" id="fl-pencil">
        ✏️ Pincel
      </button>

      <button class="fl-btn" id="fl-eraser">
        🧹 Borrador
      </button>

    </div>


    <div class="fl-tool-group">

      <label class="fl-label">

        Tamaño

        <input
          type="range"
          id="fl-brush"
          min="1"
          max="15"
          value="3">

        <span id="fl-bval">
          3
        </span>

      </label>

    </div>


    <div class="fl-tool-group">

      <button class="fl-btn" id="fl-clear">
        🗑️ Limpiar
      </button>

    </div>

  </div>


  <!-- =======================================================
       PRESETS
       ======================================================= -->

  <div class="fl-presets">

    <span class="fl-presets-label">
      Ejemplos:
    </span>

    <button
      class="fl-btn fl-preset"
      data-preset="point">
      ● Punto
    </button>

    <button
      class="fl-btn fl-preset"
      data-preset="hline">
      ━ Línea H
    </button>

    <button
      class="fl-btn fl-preset"
      data-preset="vline">
      ┃ Línea V
    </button>

    <button
      class="fl-btn fl-preset"
      data-preset="ring">
      ◯ Anillo
    </button>

    <button
      class="fl-btn fl-preset"
      data-preset="cross">
      ╋ Cruz
    </button>

  </div>

</div>


<script>
(function() {

  'use strict';


  // =========================================================
  // CONFIGURACIÓN
  // =========================================================

  var N = 256;
  var N2 = N >> 1;


  // =========================================================
  // ESTADO
  // =========================================================

  // Espacio original
  var sRe = new Float64Array(N * N);
  var sIm = new Float64Array(N * N);

  // Espacio de Fourier
  var fRe = new Float64Array(N * N);
  var fIm = new Float64Array(N * N);


  var dirty = false;

  var drawing = false;
  var erasing = false;

  var brushR = 3;

  var prevX = -1;
  var prevY = -1;


  // =========================================================
  // CANVAS
  // =========================================================

  var fCanvas =
    document.getElementById('fl-fourier');

  var sCanvas =
    document.getElementById('fl-spatial');

  var fCtx =
    fCanvas.getContext('2d');

  var sCtx =
    sCanvas.getContext('2d');


  // =========================================================
  // BUFFERS TEMPORALES PARA FFT
  // =========================================================

  var tmpRe = new Float64Array(N);
  var tmpIm = new Float64Array(N);


  // =========================================================
  // MAPEO PARA FFTSHIFT
  //
  // El resultado de la FFT tiene la frecuencia cero
  // en la esquina.
  //
  // Para visualizarlo convencionalmente queremos que
  // la frecuencia cero aparezca en el centro.
  // =========================================================

  var d2f = new Int32Array(N * N);

  for (var cy = 0; cy < N; cy++) {

    for (var cx = 0; cx < N; cx++) {

      var fx =
        ((cx - N2) + N) & (N - 1);

      var fy =
        ((cy - N2) + N) & (N - 1);

      d2f[cy * N + cx] =
        fy * N + fx;
    }
  }


  // =========================================================
  // FFT 1D
  // =========================================================

  function fft1d(re, im, n, invert) {

    var i, j, bit, t;

    // -------------------------------------------------------
    // Bit reversal
    // -------------------------------------------------------

    for (i = 1, j = 0; i < n; i++) {

      bit = n >> 1;

      while (j & bit) {

        j ^= bit;
        bit >>= 1;
      }

      j ^= bit;

      if (i < j) {

        t = re[i];
        re[i] = re[j];
        re[j] = t;

        t = im[i];
        im[i] = im[j];
        im[j] = t;
      }
    }


    // -------------------------------------------------------
    // Butterflies
    // -------------------------------------------------------

    var len;
    var ang;
    var wR;
    var wI;
    var half;
    var cR;
    var cI;
    var a;
    var b;
    var tR;
    var tI;
    var nR;


    for (len = 2; len <= n; len <<= 1) {

      ang =
        (invert ? 1 : -1)
        * 2 * Math.PI / len;

      wR = Math.cos(ang);
      wI = Math.sin(ang);


      for (i = 0; i < n; i += len) {

        cR = 1;
        cI = 0;

        half = len >> 1;


        for (j = 0; j < half; j++) {

          a = i + j;
          b = a + half;


          tR =
            re[b] * cR -
            im[b] * cI;

          tI =
            re[b] * cI +
            im[b] * cR;


          re[b] =
            re[a] - tR;

          im[b] =
            im[a] - tI;


          re[a] += tR;
          im[a] += tI;


          nR =
            cR * wR -
            cI * wI;

          cI =
            cR * wI +
            cI * wR;

          cR = nR;
        }
      }
    }


    // -------------------------------------------------------
    // Normalización de la inversa
    // -------------------------------------------------------

    if (invert) {

      for (i = 0; i < n; i++) {

        re[i] /= n;
        im[i] /= n;
      }
    }

  }


  // =========================================================
  // FFT 2D
  // =========================================================

  function computeFFT() {

    var x;
    var y;
    var off;


    // -------------------------------------------------------
    // Copiar espacio original → espacio de Fourier
    // -------------------------------------------------------

    fRe.set(sRe);
    fIm.set(sIm);


    // -------------------------------------------------------
    // FFT por filas
    // -------------------------------------------------------

    for (y = 0; y < N; y++) {

      off = y * N;


      for (x = 0; x < N; x++) {

        tmpRe[x] =
          fRe[off + x];

        tmpIm[x] =
          fIm[off + x];
      }


      fft1d(
        tmpRe,
        tmpIm,
        N,
        false
      );


      for (x = 0; x < N; x++) {

        fRe[off + x] =
          tmpRe[x];

        fIm[off + x] =
          tmpIm[x];
      }

    }


    // -------------------------------------------------------
    // FFT por columnas
    // -------------------------------------------------------

    for (x = 0; x < N; x++) {


      for (y = 0; y < N; y++) {

        tmpRe[y] =
          fRe[y * N + x];

        tmpIm[y] =
          fIm[y * N + x];
      }


      fft1d(
        tmpRe,
        tmpIm,
        N,
        false
      );


      for (y = 0; y < N; y++) {

        fRe[y * N + x] =
          tmpRe[y];

        fIm[y * N + x] =
          tmpIm[y];
      }

    }

  }


  // =========================================================
  // DIBUJAR EN EL ESPACIO ORIGINAL
  // =========================================================

  function setSpatialAt(x, y, val) {

    if (
      x < 0 ||
      x >= N ||
      y < 0 ||
      y >= N
    ) {
      return;
    }


    var index =
      y * N + x;


    sRe[index] = val;

    sIm[index] = 0;
  }


  function paintCircle(
    cx,
    cy,
    r,
    val
  ) {

    var dx;
    var dy;

    var r2 =
      r * r;


    for (
      dy = -r;
      dy <= r;
      dy++
    ) {

      for (
        dx = -r;
        dx <= r;
        dx++
      ) {

        if (
          dx * dx +
          dy * dy <= r2
        ) {

          setSpatialAt(
            cx + dx,
            cy + dy,
            val
          );
        }

      }

    }


    dirty = true;
  }


  function paintLine(
    x0,
    y0,
    x1,
    y1,
    r,
    val
  ) {

    var dx =
      x1 - x0;

    var dy =
      y1 - y0;

    var dist =
      Math.sqrt(
        dx * dx +
        dy * dy
      );

    var steps =
      Math.max(
        1,
        Math.ceil(dist)
      );


    for (
      var i = 0;
      i <= steps;
      i++
    ) {

      var t =
        i / steps;


      paintCircle(

        Math.round(
          x0 + dx * t
        ),

        Math.round(
          y0 + dy * t
        ),

        r,

        val
      );

    }

  }


  // =========================================================
  // RENDER ESPACIO ORIGINAL
  // =========================================================

  function renderSpatial() {

    var img =
      sCtx.createImageData(
        N,
        N
      );

    var d =
      img.data;


    var i;

    var minV =
      Infinity;

    var maxV =
      -Infinity;


    for (
      i = 0;
      i < N * N;
      i++
    ) {

      if (
        sRe[i] < minV
      ) {
        minV = sRe[i];
      }

      if (
        sRe[i] > maxV
      ) {
        maxV = sRe[i];
      }

    }


    var range =
      maxV - minV;


    if (
      range < 1e-12
    ) {
      range = 1;
    }


    var invRange =
      255 / range;


    var v;
    var p;


    for (
      i = 0;
      i < N * N;
      i++
    ) {

      v =
        ((sRe[i] - minV)
        * invRange) | 0;


      p =
        i << 2;


      d[p] =
        v;

      d[p + 1] =
        (v * 0.92) | 0;

      d[p + 2] =
        (v * 0.7) | 0;

      d[p + 3] =
        255;
    }


    sCtx.putImageData(
      img,
      0,
      0
    );

  }


  // =========================================================
  // RENDER ESPACIO DE FOURIER
  // =========================================================

  function renderFourier() {

    var img =
      fCtx.createImageData(
        N,
        N
      );

    var d =
      img.data;


    var mags =
      new Float64Array(
        N * N
      );


    var maxMag = 0;

    var i;
    var fi;
    var mag;


    // -------------------------------------------------------
    // Magnitud logarítmica
    // -------------------------------------------------------

    for (
      i = 0;
      i < N * N;
      i++
    ) {

      fi =
        d2f[i];


      mag =
        Math.log1p(
          Math.sqrt(
            fRe[fi] *
            fRe[fi] +

            fIm[fi] *
            fIm[fi]
          )
        );


      mags[i] =
        mag;


      if (
        mag > maxMag
      ) {

        maxMag =
          mag;
      }

    }


    var scale =
      maxMag > 0
        ? 255 / maxMag
        : 0;


    var v;
    var p;


    for (
      i = 0;
      i < N * N;
      i++
    ) {

      v =
        mags[i] *
        scale;


      p =
        i << 2;


      // Azul / cyan

      d[p] =
        (v * 0.35) | 0;

      d[p + 1] =
        (v * 0.65) | 0;

      d[p + 2] =
        v | 0;

      d[p + 3] =
        255;
    }


    fCtx.putImageData(
      img,
      0,
      0
    );

  }


  // =========================================================
  // COORDENADAS DEL MOUSE
  // =========================================================

  function getCoords(e) {

    var rect =
      sCanvas.getBoundingClientRect();


    return [

      Math.floor(
        (e.clientX - rect.left)
        / rect.width
        * N
      ),

      Math.floor(
        (e.clientY - rect.top)
        / rect.height
        * N
      )

    ];

  }


  // =========================================================
  // MOUSE
  // =========================================================

  sCanvas.addEventListener(
    'mousedown',
    function(e) {

      e.preventDefault();

      drawing = true;


      var c =
        getCoords(e);


      paintCircle(
        c[0],
        c[1],
        brushR,
        erasing ? 0 : 1
      );


      prevX =
        c[0];

      prevY =
        c[1];

    }
  );


  sCanvas.addEventListener(
    'mousemove',
    function(e) {

      if (!drawing) {
        return;
      }


      var c =
        getCoords(e);


      paintLine(

        prevX,
        prevY,

        c[0],
        c[1],

        brushR,

        erasing ? 0 : 1

      );


      prevX =
        c[0];

      prevY =
        c[1];

    }
  );


  window.addEventListener(
    'mouseup',
    function() {

      drawing = false;

      prevX = -1;
      prevY = -1;

    }
  );


  // =========================================================
  // TOUCH
  // =========================================================

  function getTouchCoords(e) {

    var touch =
      e.touches[0];


    var rect =
      sCanvas.getBoundingClientRect();


    return [

      Math.floor(
        (touch.clientX - rect.left)
        / rect.width
        * N
      ),

      Math.floor(
        (touch.clientY - rect.top)
        / rect.height
        * N
      )

    ];

  }


  sCanvas.addEventListener(
    'touchstart',
    function(e) {

      e.preventDefault();

      drawing = true;


      var c =
        getTouchCoords(e);


      paintCircle(
        c[0],
        c[1],
        brushR,
        erasing ? 0 : 1
      );


      prevX =
        c[0];

      prevY =
        c[1];

    },
    {
      passive: false
    }
  );


  sCanvas.addEventListener(
    'touchmove',
    function(e) {

      e.preventDefault();


      if (!drawing) {
        return;
      }


      var c =
        getTouchCoords(e);


      paintLine(

        prevX,
        prevY,

        c[0],
        c[1],

        brushR,

        erasing ? 0 : 1

      );


      prevX =
        c[0];

      prevY =
        c[1];

    },
    {
      passive: false
    }
  );


  sCanvas.addEventListener(
    'touchend',
    function() {

      drawing = false;

      prevX = -1;
      prevY = -1;

    }
  );


  // =========================================================
  // CONTROLES
  // =========================================================

  var pencilBtn =
    document.getElementById(
      'fl-pencil'
    );

  var eraserBtn =
    document.getElementById(
      'fl-eraser'
    );

  var brushSlider =
    document.getElementById(
      'fl-brush'
    );

  var brushValSpan =
    document.getElementById(
      'fl-bval'
    );

  var clearBtn =
    document.getElementById(
      'fl-clear'
    );


  // ---------------------------------------------------------
  // Pincel
  // ---------------------------------------------------------

  pencilBtn.addEventListener(
    'click',
    function() {

      erasing = false;

      pencilBtn.classList.add(
        'active'
      );

      eraserBtn.classList.remove(
        'active'
      );

      sCanvas.style.cursor =
        'crosshair';

    }
  );


  // ---------------------------------------------------------
  // Borrador
  // ---------------------------------------------------------

  eraserBtn.addEventListener(
    'click',
    function() {

      erasing = true;

      eraserBtn.classList.add(
        'active'
      );

      pencilBtn.classList.remove(
        'active'
      );

      sCanvas.style.cursor =
        'not-allowed';

    }
  );


  // ---------------------------------------------------------
  // Tamaño del pincel
  // ---------------------------------------------------------

  brushSlider.addEventListener(
    'input',
    function() {

      brushR =
        parseInt(
          this.value
        );

      brushValSpan.textContent =
        this.value;

    }
  );


  // ---------------------------------------------------------
  // Limpiar
  // ---------------------------------------------------------

  clearBtn.addEventListener(
    'click',
    function() {

      sRe.fill(0);
      sIm.fill(0);

      dirty = true;

    }
  );


  // =========================================================
  // PRESETS
  //
  // Ahora los presets se dibujan EN EL ESPACIO ORIGINAL.
  // =========================================================

  function clearSpatial() {

    sRe.fill(0);
    sIm.fill(0);

  }


  var presetBtns =
    document.querySelectorAll(
      '[data-preset]'
    );


  for (
    var pi = 0;
    pi < presetBtns.length;
    pi++
  ) {

    presetBtns[pi].addEventListener(
      'click',
      function() {

        var type =
          this.getAttribute(
            'data-preset'
          );


        clearSpatial();


        // ---------------------------------------------------
        // Punto
        // ---------------------------------------------------

        if (
          type === 'point'
        ) {

          paintCircle(
            N2,
            N2,
            3,
            1
          );

        }


        // ---------------------------------------------------
        // Línea horizontal
        // ---------------------------------------------------

        else if (
          type === 'hline'
        ) {

          for (
            var x = 0;
            x < N;
            x++
          ) {

            setSpatialAt(
              x,
              N2,
              1
            );

          }

        }


        // ---------------------------------------------------
        // Línea vertical
        // ---------------------------------------------------

        else if (
          type === 'vline'
        ) {

          for (
            var y = 0;
            y < N;
            y++
          ) {

            setSpatialAt(
              N2,
              y,
              1
            );

          }

        }


        // ---------------------------------------------------
        // Anillo
        // ---------------------------------------------------

        else if (
          type === 'ring'
        ) {

          var rr = 50;


          for (
            var a = 0;
            a < 360;
            a += 0.5
          ) {

            var rad =
              a *
              Math.PI /
              180;


            var rx =
              Math.round(
                N2 +
                rr *
                Math.cos(rad)
              );


            var ry =
              Math.round(
                N2 +
                rr *
                Math.sin(rad)
              );


            setSpatialAt(
              rx,
              ry,
              1
            );

          }

        }


        // ---------------------------------------------------
        // Cruz
        // ---------------------------------------------------

        else if (
          type === 'cross'
        ) {

          for (
            var x2 = 0;
            x2 < N;
            x2++
          ) {

            setSpatialAt(
              x2,
              N2,
              1
            );

          }


          for (
            var y2 = 0;
            y2 < N;
            y2++
          ) {

            setSpatialAt(
              N2,
              y2,
              1
            );

          }

        }


        dirty = true;

      }
    );

  }


  // =========================================================
  // LOOP DE ANIMACIÓN
  // =========================================================

  function animate() {

    if (dirty) {

      // Espacio original → Fourier
      computeFFT();

      // Mostrar ambos
      renderSpatial();
      renderFourier();

      dirty = false;
    }


    requestAnimationFrame(
      animate
    );

  }


  // =========================================================
  // INICIALIZACIÓN
  // =========================================================

  sRe.fill(0);
  sIm.fill(0);

  fRe.fill(0);
  fIm.fill(0);


  // Inicialmente mostramos un punto
  paintCircle(
    N2,
    N2,
    3,
    1
  );


  dirty = true;


  requestAnimationFrame(
    animate
  );

})();
</script>
"""))
```

```python
# @title
# ============================================================
# 🔬 LABORATORIO INTERACTIVO DE FOURIER
# ============================================================
# Copiá todo el contenido de este archivo en UNA celda de
# Google Colab y ejecutala. No requiere instalar nada extra.
# ============================================================

from IPython.display import HTML, display

display(HTML("""
<style>
  #fourier-lab *, #fourier-lab *::before, #fourier-lab *::after {
    box-sizing: border-box; margin: 0; padding: 0;
  }
  #fourier-lab {
    font-family: 'Segoe UI', system-ui, -apple-system, sans-serif;
    background: linear-gradient(145deg, #0d1117 0%, #161b22 50%, #0d1117 100%);
    color: #e6edf3;
    padding: 28px;
    border-radius: 18px;
    box-shadow: 0 8px 40px rgba(0,0,0,0.5);
    max-width: 920px;
    margin: 10px auto;
    border: 1px solid #21262d;
  }
  .fl-header { text-align: center; margin-bottom: 22px; }
  .fl-header h2 {
    font-size: 22px; font-weight: 700;
    background: linear-gradient(90deg, #58a6ff, #79c0ff);
    -webkit-background-clip: text; -webkit-text-fill-color: transparent;
    background-clip: text;
    margin-bottom: 6px;
  }
  .fl-header p { font-size: 13px; color: #8b949e; }

  .fl-canvases {
    display: flex; justify-content: center; align-items: flex-start;
    gap: 16px; flex-wrap: wrap; margin-bottom: 20px;
  }
  .fl-panel {
    background: rgba(22,27,34,0.85);
    border: 1px solid #30363d;
    border-radius: 14px; padding: 14px; text-align: center;
    transition: border-color 0.3s;
  }
  .fl-panel:hover { border-color: #484f58; }
  .fl-panel-title {
    font-size: 13px; font-weight: 600; margin-bottom: 10px;
    display: flex; align-items: center; justify-content: center; gap: 8px;
  }
  .fl-dot {
    width: 8px; height: 8px; border-radius: 50%; display: inline-block;
  }
  .fl-dot.blue { background: #58a6ff; box-shadow: 0 0 6px #58a6ff; }
  .fl-dot.amber { background: #d29922; box-shadow: 0 0 6px #d29922; }
  .fl-panel canvas {
    width: 340px; height: 340px; display: block; margin: 0 auto;
    border-radius: 8px; border: 2px solid #30363d;
    image-rendering: pixelated; image-rendering: -moz-crisp-edges;
    image-rendering: crisp-edges;
    background: #000;
  }
  .fourier-panel canvas {
    cursor: crosshair;
    box-shadow: 0 0 18px rgba(56,189,248,0.1);
  }
  .spatial-panel canvas {
    box-shadow: 0 0 18px rgba(210,153,34,0.1);
  }
  .fl-panel-sub {
    font-size: 11px; color: #8b949e; margin-top: 8px;
  }

  .fl-arrow {
    display: flex; flex-direction: column; align-items: center;
    justify-content: center; padding-top: 60px; gap: 4px;
  }
  .fl-arrow-label {
    font-size: 11px; color: #58a6ff; font-weight: 600;
    letter-spacing: 1px; text-transform: uppercase;
  }
  .fl-arrow-icon {
    font-size: 28px; color: #58a6ff;
    animation: fl-pulse 2s ease-in-out infinite;
  }
  @keyframes fl-pulse {
    0%, 100% { opacity: 0.5; transform: translateX(0); }
    50% { opacity: 1; transform: translateX(4px); }
  }

  .fl-toolbar {
    display: flex; justify-content: center; align-items: center;
    gap: 16px; flex-wrap: wrap; margin-bottom: 14px;
    padding: 12px 16px; background: rgba(22,27,34,0.7);
    border-radius: 10px; border: 1px solid #21262d;
  }
  .fl-tool-group {
    display: flex; align-items: center; gap: 8px;
  }
  .fl-btn {
    background: #21262d; color: #e6edf3;
    border: 1px solid #30363d; border-radius: 8px;
    padding: 7px 14px; cursor: pointer; font-size: 12px;
    transition: all 0.2s ease; font-family: inherit;
    user-select: none; white-space: nowrap;
  }
  .fl-btn:hover { background: #30363d; border-color: #58a6ff; }
  .fl-btn.active {
    background: #1f6feb; border-color: #58a6ff; color: #fff;
  }
  .fl-label {
    font-size: 12px; color: #8b949e;
    display: flex; align-items: center; gap: 6px;
  }
  .fl-label input[type=range] {
    width: 90px; accent-color: #58a6ff; cursor: pointer;
  }
  .fl-label span {
    color: #e6edf3; font-weight: 600; min-width: 18px; text-align: center;
  }

  .fl-presets {
    display: flex; justify-content: center; align-items: center;
    gap: 8px; flex-wrap: wrap;
  }
  .fl-presets-label { font-size: 12px; color: #8b949e; }
  .fl-preset {
    padding: 5px 12px; font-size: 11px;
    background: #161b22; border-color: #21262d;
  }
  .fl-preset:hover { background: #21262d; border-color: #484f58; }

  @media (max-width: 760px) {
    .fl-panel canvas { width: 280px; height: 280px; }
    .fl-arrow { padding-top: 10px; }
    .fl-arrow-icon { transform: rotate(90deg); }
    #fourier-lab { padding: 16px; }
  }
</style>

<div id="fourier-lab">
  <div class="fl-header">
    <h2>🔬 Laboratorio Interactivo de Fourier</h2>
    <p>Dibujá en el espacio de Fourier y observá el resultado en el espacio original (IFFT en tiempo real)</p>
  </div>

  <div class="fl-canvases">
    <div class="fl-panel fourier-panel">
      <div class="fl-panel-title"><span class="fl-dot blue"></span> Espacio de Fourier</div>
      <canvas id="fl-fourier" width="256" height="256"></canvas>
      <div class="fl-panel-sub">✏️ Dibujá aquí con el mouse</div>
    </div>
    <div class="fl-arrow">
      <div class="fl-arrow-label">IFFT</div>
      <div class="fl-arrow-icon">→</div>
    </div>
    <div class="fl-panel spatial-panel">
      <div class="fl-panel-title"><span class="fl-dot amber"></span> Espacio Original</div>
      <canvas id="fl-spatial" width="256" height="256"></canvas>
      <div class="fl-panel-sub">Transformada inversa (resultado automático)</div>
    </div>
  </div>

  <div class="fl-toolbar">
    <div class="fl-tool-group">
      <button class="fl-btn active" id="fl-pencil">✏️ Pincel</button>
      <button class="fl-btn" id="fl-eraser">🧹 Borrador</button>
    </div>
    <div class="fl-tool-group">
      <label class="fl-label">
        Tamaño
        <input type="range" id="fl-brush" min="1" max="15" value="3">
        <span id="fl-bval">3</span>
      </label>
    </div>
    <div class="fl-tool-group">
      <button class="fl-btn" id="fl-clear">🗑️ Limpiar</button>
    </div>
  </div>

  <div class="fl-presets">
    <span class="fl-presets-label">Ejemplos:</span>
    <button class="fl-btn fl-preset" data-preset="point">● Punto</button>
    <button class="fl-btn fl-preset" data-preset="hline">━ Línea H</button>
    <button class="fl-btn fl-preset" data-preset="vline">┃ Línea V</button>
    <button class="fl-btn fl-preset" data-preset="ring">◯ Anillo</button>
    <button class="fl-btn fl-preset" data-preset="cross">╋ Cruz</button>
  </div>
</div>

<script>
(function() {
  'use strict';

  // ==================== CONFIGURACION ====================
  var N = 256;
  var N2 = N >> 1;  // 128

  // ==================== ESTADO ====================
  var fRe = new Float64Array(N * N);  // freq domain real
  var fIm = new Float64Array(N * N);  // freq domain imag
  var sRe = new Float64Array(N * N);  // spatial domain real (resultado IFFT)
  var sIm = new Float64Array(N * N);  // spatial domain imag

  var dirty = false;
  var drawing = false;
  var erasing = false;
  var brushR = 3;
  var prevX = -1, prevY = -1;

  // Canvas refs
  var fCanvas = document.getElementById('fl-fourier');
  var sCanvas = document.getElementById('fl-spatial');
  var fCtx = fCanvas.getContext('2d');
  var sCtx = sCanvas.getContext('2d');

  // Precompute display→frequency index mapping (fftshift inverso)
  var d2f = new Int32Array(N * N);
  for (var cy = 0; cy < N; cy++) {
    for (var cx = 0; cx < N; cx++) {
      var fx = ((cx - N2) + N) & (N - 1);
      var fy = ((cy - N2) + N) & (N - 1);
      d2f[cy * N + cx] = fy * N + fx;
    }
  }

  // ==================== FFT 1D (Cooley-Tukey iterativo) ====================
  function fft1d(re, im, n, invert) {
    // Bit-reversal permutation
    var i, j, bit, t;
    for (i = 1, j = 0; i < n; i++) {
      bit = n >> 1;
      while (j & bit) { j ^= bit; bit >>= 1; }
      j ^= bit;
      if (i < j) {
        t = re[i]; re[i] = re[j]; re[j] = t;
        t = im[i]; im[i] = im[j]; im[j] = t;
      }
    }

    // Butterflies
    var len, ang, wR, wI, half, cR, cI, a, b, tR, tI, nR;
    for (len = 2; len <= n; len <<= 1) {
      ang = (invert ? 1 : -1) * 2 * Math.PI / len;
      wR = Math.cos(ang);
      wI = Math.sin(ang);
      for (i = 0; i < n; i += len) {
        cR = 1; cI = 0;
        half = len >> 1;
        for (j = 0; j < half; j++) {
          a = i + j;
          b = a + half;
          tR = re[b] * cR - im[b] * cI;
          tI = re[b] * cI + im[b] * cR;
          re[b] = re[a] - tR;
          im[b] = im[a] - tI;
          re[a] += tR;
          im[a] += tI;
          nR = cR * wR - cI * wI;
          cI = cR * wI + cI * wR;
          cR = nR;
        }
      }
    }

    // Escalar para inversa
    if (invert) {
      for (i = 0; i < n; i++) {
        re[i] /= n;
        im[i] /= n;
      }
    }
  }

  // ==================== IFFT 2D ====================
  var tmpRe = new Float64Array(N);
  var tmpIm = new Float64Array(N);

  function computeIFFT() {
    var x, y, off;
    // Copiar freq → spatial
    sRe.set(fRe);
    sIm.set(fIm);

    // IFFT por filas
    for (y = 0; y < N; y++) {
      off = y * N;
      for (x = 0; x < N; x++) {
        tmpRe[x] = sRe[off + x];
        tmpIm[x] = sIm[off + x];
      }
      fft1d(tmpRe, tmpIm, N, true);
      for (x = 0; x < N; x++) {
        sRe[off + x] = tmpRe[x];
        sIm[off + x] = tmpIm[x];
      }
    }

    // IFFT por columnas
    for (x = 0; x < N; x++) {
      for (y = 0; y < N; y++) {
        tmpRe[y] = sRe[y * N + x];
        tmpIm[y] = sIm[y * N + x];
      }
      fft1d(tmpRe, tmpIm, N, true);
      for (y = 0; y < N; y++) {
        sRe[y * N + x] = tmpRe[y];
        sIm[y * N + x] = tmpIm[y];
      }
    }
  }

  // ==================== DIBUJO ====================
  function setFreqAt(cx, cy, val) {
    if (cx < 0 || cx >= N || cy < 0 || cy >= N) return;
    var fi = d2f[cy * N + cx];
    fRe[fi] = val;
    fIm[fi] = 0;
    // Simetria hermitiana: conjugado en posicion espejo
    var mcx = (N - cx) & (N - 1);
    var mcy = (N - cy) & (N - 1);
    var mfi = d2f[mcy * N + mcx];
    fRe[mfi] = val;
    fIm[mfi] = 0;
  }

  function paintCircle(cx, cy, r, val) {
    var dx, dy, r2 = r * r;
    for (dy = -r; dy <= r; dy++) {
      for (dx = -r; dx <= r; dx++) {
        if (dx * dx + dy * dy <= r2) {
          setFreqAt(cx + dx, cy + dy, val);
        }
      }
    }
    dirty = true;
  }

  function paintLine(x0, y0, x1, y1, r, val) {
    var dx = x1 - x0, dy = y1 - y0;
    var dist = Math.sqrt(dx * dx + dy * dy);
    var steps = Math.max(1, Math.ceil(dist));
    var i, t;
    for (i = 0; i <= steps; i++) {
      t = i / steps;
      paintCircle(
        Math.round(x0 + dx * t),
        Math.round(y0 + dy * t),
        r, val
      );
    }
  }

  // ==================== RENDERIZADO ====================
  function renderFourier() {
    var img = fCtx.createImageData(N, N);
    var d = img.data;
    var i, fi, mag, maxMag = 0;
    var mags = new Float64Array(N * N);

    for (i = 0; i < N * N; i++) {
      fi = d2f[i];  // el mapeo ya esta hecho por fila (i = cy*N + cx)
      // Hmm, d2f esta indexado como [cy*N+cx], y i va 0..N*N-1
      // Para renderizar, i = cy*N+cx (recorremos canvas en orden), perfecto.
      mag = Math.log1p(Math.sqrt(fRe[fi] * fRe[fi] + fIm[fi] * fIm[fi]));
      mags[i] = mag;
      if (mag > maxMag) maxMag = mag;
    }

    var scale = maxMag > 0 ? 255 / maxMag : 0;
    var v, p;
    for (i = 0; i < N * N; i++) {
      v = mags[i] * scale;
      p = i << 2;
      // Tinte azul/cyan
      d[p]     = (v * 0.35) | 0;
      d[p + 1] = (v * 0.65) | 0;
      d[p + 2] = v | 0;
      d[p + 3] = 255;
    }

    fCtx.putImageData(img, 0, 0);
  }

  function renderSpatial() {
    var img = sCtx.createImageData(N, N);
    var d = img.data;
    var i, minV = Infinity, maxV = -Infinity;

    for (i = 0; i < N * N; i++) {
      if (sRe[i] < minV) minV = sRe[i];
      if (sRe[i] > maxV) maxV = sRe[i];
    }

    var range = maxV - minV;
    if (range < 1e-12) range = 1;
    var invRange = 255 / range;
    var v, p;

    for (i = 0; i < N * N; i++) {
      v = ((sRe[i] - minV) * invRange) | 0;
      p = i << 2;
      // Tinte ambar/calido
      d[p]     = v;
      d[p + 1] = (v * 0.92) | 0;
      d[p + 2] = (v * 0.7) | 0;
      d[p + 3] = 255;
    }

    sCtx.putImageData(img, 0, 0);
  }

  // ==================== EVENTOS DEL MOUSE ====================
  function getCoords(e) {
    var rect = fCanvas.getBoundingClientRect();
    return [
      Math.floor((e.clientX - rect.left) / rect.width * N),
      Math.floor((e.clientY - rect.top) / rect.height * N)
    ];
  }

  fCanvas.addEventListener('mousedown', function(e) {
    e.preventDefault();
    drawing = true;
    var c = getCoords(e);
    paintCircle(c[0], c[1], brushR, erasing ? 0 : 1);
    prevX = c[0]; prevY = c[1];
  });

  fCanvas.addEventListener('mousemove', function(e) {
    if (!drawing) return;
    var c = getCoords(e);
    paintLine(prevX, prevY, c[0], c[1], brushR, erasing ? 0 : 1);
    prevX = c[0]; prevY = c[1];
  });

  window.addEventListener('mouseup', function() {
    drawing = false;
    prevX = prevY = -1;
  });

  // ==================== EVENTOS TOUCH ====================
  function getTouchCoords(e) {
    var touch = e.touches[0];
    var rect = fCanvas.getBoundingClientRect();
    return [
      Math.floor((touch.clientX - rect.left) / rect.width * N),
      Math.floor((touch.clientY - rect.top) / rect.height * N)
    ];
  }

  fCanvas.addEventListener('touchstart', function(e) {
    e.preventDefault();
    drawing = true;
    var c = getTouchCoords(e);
    paintCircle(c[0], c[1], brushR, erasing ? 0 : 1);
    prevX = c[0]; prevY = c[1];
  }, {passive: false});

  fCanvas.addEventListener('touchmove', function(e) {
    e.preventDefault();
    if (!drawing) return;
    var c = getTouchCoords(e);
    paintLine(prevX, prevY, c[0], c[1], brushR, erasing ? 0 : 1);
    prevX = c[0]; prevY = c[1];
  }, {passive: false});

  fCanvas.addEventListener('touchend', function() {
    drawing = false;
    prevX = prevY = -1;
  });

  // ==================== CONTROLES ====================
  var pencilBtn = document.getElementById('fl-pencil');
  var eraserBtn = document.getElementById('fl-eraser');
  var brushSlider = document.getElementById('fl-brush');
  var brushValSpan = document.getElementById('fl-bval');
  var clearBtn = document.getElementById('fl-clear');

  pencilBtn.addEventListener('click', function() {
    erasing = false;
    pencilBtn.classList.add('active');
    eraserBtn.classList.remove('active');
    fCanvas.style.cursor = 'crosshair';
  });

  eraserBtn.addEventListener('click', function() {
    erasing = true;
    eraserBtn.classList.add('active');
    pencilBtn.classList.remove('active');
    fCanvas.style.cursor = 'not-allowed';
  });

  brushSlider.addEventListener('input', function() {
    brushR = parseInt(this.value);
    brushValSpan.textContent = this.value;
  });

  clearBtn.addEventListener('click', function() {
    fRe.fill(0);
    fIm.fill(0);
    dirty = true;
  });

  // ==================== PRESETS ====================
  function clearFreq() {
    fRe.fill(0);
    fIm.fill(0);
  }

  var presetBtns = document.querySelectorAll('[data-preset]');
  for (var pi = 0; pi < presetBtns.length; pi++) {
    presetBtns[pi].addEventListener('click', function() {
      var type = this.getAttribute('data-preset');
      clearFreq();

      if (type === 'point') {
        // Un punto desplazado horizontalmente → onda coseno vertical
        setFreqAt(N2 + 20, N2, 1);

      } else if (type === 'hline') {
        // Linea horizontal en freq → bordes/rayas verticales
        for (var x = 0; x < N; x++) setFreqAt(x, N2, 1);

      } else if (type === 'vline') {
        // Linea vertical en freq → bordes/rayas horizontales
        for (var y = 0; y < N; y++) setFreqAt(N2, y, 1);

      } else if (type === 'ring') {
        // Anillo en freq → patron radial
        var rr = 30;
        for (var a = 0; a < 360; a += 0.5) {
          var rx = Math.round(N2 + rr * Math.cos(a * Math.PI / 180));
          var ry = Math.round(N2 + rr * Math.sin(a * Math.PI / 180));
          setFreqAt(rx, ry, 1);
        }

      } else if (type === 'cross') {
        // Cruz → rejilla en el espacio original
        for (var x2 = 0; x2 < N; x2++) setFreqAt(x2, N2, 1);
        for (var y2 = 0; y2 < N; y2++) setFreqAt(N2, y2, 1);
      }

      dirty = true;
    });
  }

  // ==================== LOOP DE ANIMACION ====================
  function animate() {
    if (dirty) {
      computeIFFT();
      renderFourier();
      renderSpatial();
      dirty = false;
    }
    requestAnimationFrame(animate);
  }

  // Render inicial (todo negro)
  renderFourier();
  renderSpatial();

  // Inicializar con el preset "punto" para que se vea algo interesante
  setFreqAt(N2 + 20, N2, 1);
  dirty = true;

  requestAnimationFrame(animate);
})();
</script>
"""))
```
